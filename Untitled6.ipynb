```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
sentiment_path = "Sentiment_Classification/amazon_distilbert"

# Load tokenizer
sentiment_tokenizer = AutoTokenizer.from_pretrained(
    sentiment_path,
    local_files_only=True
)

# Load model
sentiment_model = AutoModelForSequenceClassification.from_pretrained(
    sentiment_path,
    local_files_only=True
)

# Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

sentiment_model.to(device)
sentiment_model.eval()

print("✅ Sentiment model loaded successfully!")
print("Device:", device)
```

```python
print("Sentiment labels:")
print(sentiment_model.config.id2label)
```

```python
def predict_intent(text):
    """
    Predict the customer-support intent.
    """
    intent = intent_model.predict([text])[0]
    return intent


def predict_sentiment(text):
    """
    Predict sentiment and return its confidence.
    """
    inputs = sentiment_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    # Move inputs to the same device as the model
    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():
        outputs = sentiment_model(**inputs)
        probabilities = torch.softmax(outputs.logits, dim=-1)

    prediction = torch.argmax(probabilities, dim=-1).item()
    confidence = probabilities[0][prediction].item()

    sentiment = sentiment_model.config.id2label[prediction]

    return sentiment, confidence
```

```python
def intent_guardrail(text, predicted_intent):

    text_lower = text.lower()

    # ==========================================
    # 1. Password / login
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "forgot my password",
        "forgot password",
        "reset my password",
        "can't log in",
        "cannot log in",
        "can't login",
        "cannot login",
        "unable to log in"
    ]):
        return "recover_password"

    # ==========================================
    # 2. Cancel order
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "cancel my order",
        "cancel the order",
        "cancel order"
    ]):
        return "cancel_order"

    # ==========================================
    # 3. Track order
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "where is my order",
        "track my order",
        "track the order",
        "order status",
        "where's my order"
    ]):
        return "track_order"

    # ==========================================
    # 4. Refund
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "want a refund",
        "get a refund",
        "request a refund",
        "refund my order"
    ]):
        return "get_refund"

    # ==========================================
    # 5. Payment
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "payment problem",
        "payment issue",
        "payment failed",
        "payment doesn't work",
        "payment does not work"
    ]):
        return "payment_issue"

    # ==========================================
    # 6. Human agent
    # ==========================================
    if any(phrase in text_lower for phrase in [
        "talk to a human",
        "speak to a human",
        "human agent",
        "real person",
        "customer service agent"
    ]):
        return "contact_human_agent"

    # ==========================================
    # 7. Damaged / defective / wrong product
    # ==========================================
    if any(word in text_lower for word in [
        "damaged",
        "broken",
        "defective",
        "wrong item",
        "wrong product"
    ]):
        return "complaint"

    # ==========================================
    # 8. Generic positive/negative feedback
    # ==========================================
    feedback_words = [
        "happy",
        "satisfied",
        "great",
        "excellent",
        "love",
        "amazing",
        "terrible",
        "awful",
        "disappointed"
    ]

    if any(word in text_lower for word in feedback_words):

        # If there is no specific support intent,
        # treat it as general feedback.
        support_keywords = [
            "password",
            "order",
            "refund",
            "payment",
            "shipping",
            "delivery",
            "invoice",
            "account"
        ]

        if not any(word in text_lower for word in support_keywords):
            return "general_feedback"

    # ==========================================
    # 9. Otherwise keep ML prediction
    # ==========================================
    return predicted_intent
```

```python
def analyze_message(text):

    # 1. ML intent prediction
    predicted_intent = predict_intent(text)

    # 2. Intent guardrail
    intent = intent_guardrail(
        text,
        predicted_intent
    )

    # 3. Sentiment prediction
    sentiment, confidence = predict_sentiment(text)

    return {
        "text": text,
        "intent": intent,
        "original_intent": predicted_intent,
        "sentiment": sentiment,
        "sentiment_confidence": round(confidence, 4)
    }
```

```python
# ==========================================
# PRIORITY ENGINE
# ==========================================

HIGH_INTENTS = {
    "complaint",
    "payment_issue",
    "get_refund",
    "track_refund",
    "cancel_order",
    "contact_human_agent",
    "contact_customer_service"
}

MEDIUM_INTENTS = {
    "track_order",
    "change_shipping_address",
    "change_order",
    "delivery_period",
    "delivery_options",
    "check_cancellation_fee"
}


def calculate_priority(intent, sentiment):

    # -------------------------
    # 1. Base priority
    # -------------------------

    if intent in HIGH_INTENTS:
        priority = 3

    elif intent in MEDIUM_INTENTS:
        priority = 2

    else:
        priority = 1

    # -------------------------
    # 2. Negative sentiment
    # -------------------------

    if sentiment == "negative":
        priority = min(priority + 1, 3)

    # -------------------------
    # 3. Convert to label
    # -------------------------

    priority_labels = {
        1: "LOW",
        2: "MEDIUM",
        3: "HIGH"
    }

    return priority_labels[priority]
```

```python
def analyze_message(text):

    # 1. ML intent
    predicted_intent = predict_intent(text)

    # 2. Guardrail
    intent = intent_guardrail(
        text,
        predicted_intent
    )

    # 3. Sentiment
    sentiment, confidence = predict_sentiment(text)

    # 4. Priority
    priority = calculate_priority(
        intent,
        sentiment
    )

    return {
        "text": text,
        "original_intent": predicted_intent,
        "intent": intent,
        "sentiment": sentiment,
        "sentiment_confidence": round(confidence, 4),
        "priority": priority
    }
```

```python
def chat(message):

    global conversation_state

    # =====================================
    # 1. Check if we are waiting for order ID
    # =====================================

    if conversation_state["waiting_for"] == "order_id":

        order_id = extract_order_id(message)

        if order_id is not None:

            intent = conversation_state["last_intent"]

            conversation_state["waiting_for"] = None

            # Track order
            if intent == "track_order":

                order = get_order_status(order_id)

                if order is None:
                    return f"I couldn't find order {order_id}. Please check the order ID."

                return (
                    f"Your order {order['order_id']} for "
                    f"{order['product']} is currently "
                    f"{order['status']}. "
                    f"Expected delivery: {order['delivery_date']}."
                )

            # Track refund
            if intent == "track_refund":

                refund = get_refund_status(order_id)

                if refund is None:
                    return f"I couldn't find a refund for order {order_id}."

                return (
                    f"Your refund {refund['refund_id']} is currently "
                    f"{refund['status']}. "
                    f"Amount: ${refund['amount']:.2f}."
                )

        else:

            return "Please provide a valid order ID, such as ORD001."


    # =====================================
    # 2. Analyze new message
    # =====================================

    analysis = analyze_message(message)

    intent = analysis["intent"]

    # =====================================
    # 3. Intents requiring an order ID
    # =====================================

    if intent in [
        "track_order",
        "track_refund"
    ]:

        conversation_state["waiting_for"] = "order_id"
        conversation_state["last_intent"] = intent

        if intent == "track_order":
            return "Sure! Please provide your order ID, for example ORD001."

        if intent == "track_refund":
            return "Sure! Please provide your order ID so I can check your refund."


    # =====================================
    # 4. Normal response
    # =====================================

    return generate_response(intent)
```

```python
%%writefile app.py

import streamlit as st
import re
import sqlite3
import joblib
import torch

from transformers import AutoTokenizer, AutoModelForSequenceClassification


# ============================================================
# CONFIGURATION
# ============================================================

DB_PATH = "/content/drive/MyDrive/FogProject/customer_support.db"

INTENT_PATH = (
    "/content/drive/MyDrive/FogProject/"
    "Intent_Classification/intent_classifier.pkl"
)

SENTIMENT_PATH = (
    "/content/drive/MyDrive/FogProject/"
    "Sentiment_Classification/amazon_distilbert"
)


# ============================================================
# DATABASE
# ============================================================

conn = sqlite3.connect(
    DB_PATH,
    check_same_thread=False
)

cursor = conn.cursor()


def get_order_status(order_id):

    cursor.execute("""
        SELECT order_id, product, status, delivery_date, amount
        FROM orders
        WHERE order_id = ?
    """, (order_id,))

    order = cursor.fetchone()

    if order is None:
        return None

    return {
        "order_id": order[0],
        "product": order[1],
        "status": order[2],
        "delivery_date": order[3],
        "amount": order[4]
    }


def get_refund_status(order_id):

    cursor.execute("""
        SELECT refund_id, status, amount
        FROM refunds
        WHERE order_id = ?
    """, (order_id,))

    refund = cursor.fetchone()

    if refund is None:
        return None

    return {
        "refund_id": refund[0],
        "status": refund[1],
        "amount": refund[2]
    }


# ============================================================
# LOAD INTENT MODEL
# ============================================================

intent_model = joblib.load(INTENT_PATH)


# ============================================================
# LOAD SENTIMENT MODEL
# ============================================================

sentiment_tokenizer = AutoTokenizer.from_pretrained(
    SENTIMENT_PATH,
    local_files_only=True
)

sentiment_model = AutoModelForSequenceClassification.from_pretrained(
    SENTIMENT_PATH,
    local_files_only=True
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

sentiment_model.to(device)
sentiment_model.eval()


# ============================================================
# INTENT
# ============================================================

def predict_intent(text):

    return intent_model.predict([text])[0]


def intent_guardrail(text, predicted_intent):

    text_lower = text.lower()

    if any(phrase in text_lower for phrase in [
        "forgot my password",
        "forgot password",
        "reset my password",
        "can't log in",
        "cannot log in",
        "can't login",
        "cannot login",
        "unable to log in"
    ]):
        return "recover_password"

    if any(phrase in text_lower for phrase in [
        "cancel my order",
        "cancel the order",
        "cancel order"
    ]):
        return "cancel_order"

    if any(phrase in text_lower for phrase in [
        "where is my order",
        "track my order",
        "track the order",
        "order status",
        "where's my order"
    ]):
        return "track_order"

    if any(phrase in text_lower for phrase in [
        "want a refund",
        "get a refund",
        "request a refund",
        "refund my order"
    ]):
        return "get_refund"

    if any(phrase in text_lower for phrase in [
        "payment problem",
        "payment issue",
        "payment failed",
        "payment doesn't work",
        "payment does not work"
    ]):
        return "payment_issue"

    if any(phrase in text_lower for phrase in [
        "talk to a human",
        "speak to a human",
        "human agent",
        "real person",
        "customer service agent"
    ]):
        return "contact_human_agent"

    if any(word in text_lower for word in [
        "damaged",
        "broken",
        "defective",
        "wrong item",
        "wrong product"
    ]):
        return "complaint"

    feedback_words = [
        "happy",
        "satisfied",
        "great",
        "excellent",
        "love",
        "amazing",
        "terrible",
        "awful",
        "disappointed"
    ]

    if any(word in text_lower for word in feedback_words):

        support_keywords = [
            "password",
            "order",
            "refund",
            "payment",
            "shipping",
            "delivery",
            "invoice",
            "account"
        ]

        if not any(word in text_lower for word in support_keywords):
            return "general_feedback"

    return predicted_intent


# ============================================================
# SENTIMENT
# ============================================================

def predict_sentiment(text):

    inputs = sentiment_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = sentiment_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )

    prediction = torch.argmax(
        probabilities,
        dim=-1
    ).item()

    confidence = probabilities[
        0
    ][prediction].item()

    sentiment = sentiment_model.config.id2label[
        prediction
    ]

    return sentiment, confidence


# ============================================================
# PRIORITY
# ============================================================

HIGH_INTENTS = {
    "complaint",
    "payment_issue",
    "get_refund",
    "track_refund",
    "cancel_order",
    "contact_human_agent",
    "contact_customer_service"
}

MEDIUM_INTENTS = {
    "track_order",
    "change_shipping_address",
    "change_order",
    "delivery_period",
    "delivery_options",
    "check_cancellation_fee"
}


def calculate_priority(intent, sentiment):

    if intent in HIGH_INTENTS:
        priority = 3

    elif intent in MEDIUM_INTENTS:
        priority = 2

    else:
        priority = 1

    if sentiment == "negative":
        priority = min(
            priority + 1,
            3
        )

    return {
        1: "LOW",
        2: "MEDIUM",
        3: "HIGH"
    }[priority]


# ============================================================
# ANALYSIS
# ============================================================

def analyze_message(text):

    predicted_intent = predict_intent(text)

    intent = intent_guardrail(
        text,
        predicted_intent
    )

    sentiment, confidence = predict_sentiment(text)

    priority = calculate_priority(
        intent,
        sentiment
    )

    return {
        "intent": intent,
        "sentiment": sentiment,
        "sentiment_confidence": round(
            confidence,
            4
        ),
        "priority": priority
    }


# ============================================================
# RESPONSE ENGINE
# ============================================================

def generate_response(intent):

    responses = {

        "cancel_order":
            "I can help you cancel your order. "
            "Please provide your order ID.",

        "get_refund":
            "I can help you with your refund. "
            "Please provide your order ID.",

        "payment_issue":
            "I'm sorry you're experiencing a payment issue. "
            "Please tell me what happened or provide your order ID.",

        "recover_password":
            "I can help you recover your account password. "
            "Please follow the password reset process.",

        "change_shipping_address":
            "I can help you change your shipping address. "
            "Please provide your order ID.",

        "change_order":
            "I can help you modify your order. "
            "Please provide your order ID.",

        "delivery_period":
            "I can help you check the expected delivery period "
            "for your order.",

        "delivery_options":
            "I can provide information about the available "
            "delivery options.",

        "check_cancellation_fee":
            "I can check whether a cancellation fee applies "
            "to your order.",

        "check_refund_policy":
            "I can explain our refund policy and help you "
            "determine whether your order is eligible.",

        "complaint":
            "I'm sorry about the problem with your order. "
            "Please provide your order ID so I can help "
            "resolve the issue.",

        "contact_human_agent":
            "I'll help you get in touch with a human "
            "customer-support agent.",

        "contact_customer_service":
            "I can help you contact customer support.",

        "general_feedback":
            "Thank you for your feedback! "
            "We appreciate hearing from you."
    }

    return responses.get(
        intent,
        "I'm here to help. Could you provide more details?"
    )


# ============================================================
# ORDER ID
# ============================================================

def extract_order_id(text):

    match = re.search(
        r"\bORD\d+\b",
        text.upper()
    )

    if match:
        return match.group()

    return None


# ============================================================
# CONVERSATION STATE
# ============================================================

if "waiting_for" not in st.session_state:
    st.session_state.waiting_for = None

if "last_intent" not in st.session_state:
    st.session_state.last_intent = None


# ============================================================
# CHAT FUNCTION
# ============================================================

def chat(message):

    # User is providing an order ID
    if st.session_state.waiting_for == "order_id":

        order_id = extract_order_id(message)

        if order_id is None:
            return (
                "Please provide a valid order ID, "
                "such as ORD001."
            )

        intent = st.session_state.last_intent

        st.session_state.waiting_for = None
        st.session_state.last_intent = None

        # Track order
        if intent == "track_order":

            order = get_order_status(order_id)

            if order is None:
                return (
                    f"I couldn't find order {order_id}. "
                    "Please check the order ID."
                )

            return (
                f"Your order {order['order_id']} for "
                f"{order['product']} is currently "
                f"{order['status']}. "
                f"Expected delivery: "
                f"{order['delivery_date']}."
            )

        # Track refund
        if intent == "track_refund":

            refund = get_refund_status(order_id)

            if refund is None:
                return (
                    f"I couldn't find a refund for "
                    f"order {order_id}."
                )

            return (
                f"Your refund {refund['refund_id']} "
                f"is currently {refund['status']}. "
                f"Amount: ${refund['amount']:.2f}."
            )

    # Analyze new message
    analysis = analyze_message(message)

    intent = analysis["intent"]

    # Need order ID
    if intent in [
        "track_order",
        "track_refund"
    ]:

        st.session_state.waiting_for = "order_id"
        st.session_state.last_intent = intent

        if intent == "track_order":
            return (
                "Sure! Please provide your order ID, "
                "for example ORD001."
            )

        return (
            "Sure! Please provide your order ID "
            "so I can check your refund."
        )

    return generate_response(intent)


# ============================================================
# STREAMLIT UI
# ============================================================

st.set_page_config(
    page_title="Customer Support AI",
    page_icon="🤖",
    layout="centered"
)

st.title("🤖 Customer Support AI")

st.caption(
    "Intent Classification • Sentiment Analysis • "
    "Priority Detection • Database Tools"
)


# Display previous messages
for message in st.session_state.get(
    "messages",
    []
):

    with st.chat_message(message["role"]):
        st.write(message["content"])


# User input
user_message = st.chat_input(
    "How can I help you?"
)


if user_message:

    if "messages" not in st.session_state:
        st.session_state.messages = []

    # Display user message
    st.session_state.messages.append({
        "role": "user",
        "content": user_message
    })

    # Generate response
    response = chat(user_message)

    # Store response
    st.session_state.messages.append({
        "role": "assistant",
        "content": response
    })

    st.rerun()
```
